## Problem 1: Whitespace tokenize and count

```python
s = "Attention is all you need"

tokens = s.lower().split()

print("tokens:", tokens)
print("count:", len(tokens))
```

## Problem 2: Two tokenizers, one word

```python
word = "playing"

whitespace = word.split()
suffix = ["play", "##ing"]

print("whitespace:", whitespace)
print("suffix:", suffix)
```

## Problem 3: Encode to ids and decode back

```python
vocab = {
    "i": 0,
    "love": 1,
    "nlp": 2
}

tokens = ["i", "love", "nlp", "love"]

ids = []

for token in tokens:
    ids.append(vocab[token])

id_to_token = {}

for token, token_id in vocab.items():
    id_to_token[token_id] = token

decoded = []

for token_id in ids:
    decoded.append(id_to_token[token_id])

print("ids:", ids)
print("decoded:", decoded)
```

## Problem 4: Token count is not word count

```python
text = "cats running"
tokens = ["cat", "##s", "runn", "##ing"]

words = text.split()

print("words:", len(words))
print("tokens:", len(tokens))
```

## Problem 5: A bag of tokens loses order

```python
a = "dog bites man"
b = "man bites dog"

tokens_a = a.split()
tokens_b = b.split()

same_bag = sorted(tokens_a) == sorted(tokens_b)
same_sentence = a == b

print("same bag:", same_bag)
print("same sentence:", same_sentence)
```

## Problem 6: Why order matters

```python
print(
    "a bag of independent tokens keeps which words appear but not their order, "
    "so it cannot tell 'dog bites man' from 'man bites dog'; attention restores order and context"
)
```
