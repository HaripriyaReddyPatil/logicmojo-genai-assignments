## Problem 1: Feed-forward parameter count

```python
d = 8

first_layer_weights = d * (4 * d)
first_layer_biases = 4 * d

second_layer_weights = (4 * d) * d
second_layer_biases = d

total_params = (
    first_layer_weights
    + first_layer_biases
    + second_layer_weights
    + second_layer_biases
)

print(f"feed-forward params: {total_params}")
```

## Problem 2: Head dimension

```python
d_model = 12
n_heads = 4

head_dim = d_model // n_heads

print(f"per-head dimension: {head_dim}")
```

## Problem 3: A residual connection

```python
import numpy as np

x = np.array([1., 2., 3.])
layer_out = np.array([0.1, 0.1, 0.1])

residual = x + layer_out

print("residual:", residual.tolist())
```

## Problem 4: Concatenate two heads

```python
h1 = np.array([1., 2.])
h2 = np.array([3., 4.])

combined = np.concatenate([h1, h2])

print("combined:", combined.tolist())
```

## Problem 5: LayerNorm centering

```python
v = np.array([2., 4., 6.])

mean = v.mean()
centered = v - mean

print("mean:", mean)
print("centered:", centered.tolist())
```
