## Problem 1: Read a JSON response

```python
response = {
    "city": "Paris",
    "temp_c": 18,
    "conditions": "cloudy"
}

city = response["city"]
temperature = response["temp_c"]
conditions = response["conditions"]

print(f"{city}: {temperature}C, {conditions}")

humidity = response.get("humidity", "n/a")
print(f"humidity: {humidity}")
```

## Problem 2: Check the status code

```python
responses = [
    {"status_code": 200, "data": {"answer": 42}},
    {"status_code": 404, "data": None}
]

for response in responses:
    status = response["status_code"]

    if status == 200:
        print(f"ok: {response['data']}")
    else:
        print(f"request failed with status {status}")
```

## Problem 3: Call a local model with Ollama

```python
import requests

url = "http://localhost:11434/api/generate"

request_data = {
    "model": "llama3.2",
    "prompt": "What's NATO? Give a brief explanation.",
    "stream": False
}

response = requests.post(url, json=request_data)

result = response.json()
reply = result["response"]

print(reply)
```

## Problem 4: Call a hosted model with Groq

```python
import os
import requests
import getpass

# Ask for the API key safely if it is not already set
if not os.environ.get("GROQ_API_KEY"):
    os.environ["GROQ_API_KEY"] = getpass.getpass("Enter your Groq API key: ")

url = "https://api.groq.com/openai/v1/chat/completions"

headers = {
    "Authorization": f"Bearer {os.environ['GROQ_API_KEY']}",
    "Content-Type": "application/json"
}

request_data = {
    "model": "openai/gpt-oss-20b",
    "messages": [
        {
            "role": "user",
            "content": "What's BRICS? Give a brief explanation."
        }
    ]
}

response = requests.post(
    url,
    headers=headers,
    json=request_data,
    timeout=30
)

if response.status_code == 200:
    result = response.json()
    reply = result["choices"][0]["message"]["content"]
    print("groq:", reply)
else:
    print(f"request failed with status {response.status_code}")
    print(response.json())
```
