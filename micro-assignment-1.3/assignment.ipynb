## Problem 1: An Account class

```python
class Account:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount


account_a = Account(100)
account_b = Account()

account_a.deposit(50)
account_b.deposit(20)

print(f"Account A balance: {account_a.balance}")
print(f"Account B balance: {account_b.balance}")
```

## Problem 2: Sum the valid numbers

```python
raw = ["10", "7", "x", "3"]

total = 0
skipped = 0

for item in raw:
    try:
        number = int(item)
        total += number
    except ValueError:
        skipped += 1

print(f"total: {total}")
print(f"skipped: {skipped}")
```

## Problem 3: JSON is a dict in text form

```python
import json

data = {"course": "GenAI", "classes": 45}

json_text = json.dumps(data)
new_data = json.loads(json_text)

new_data["classes"] += 1

print(f"as text: {json_text}")
print(f"updated classes: {new_data['classes']}")
```

## Problem 4: A savings subclass

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def summary(self):
        return f"Account: {self.balance}"


class SavingsAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate

    def summary(self):
        return f"SavingsAccount: {self.balance} at {self.rate}"


accounts = [
    Account(100),
    SavingsAccount(200, 0.05)
]

for account in accounts:
    print(account.summary())
```

## Problem 5: Missing file, handled

```python
import json

filename = "missing_data.json"

try:
    with open(filename) as file:
        records = json.load(file)

except FileNotFoundError:
    print(f"could not find {filename}; starting empty")
    records = []

print(f"records: {len(records)}")
```
