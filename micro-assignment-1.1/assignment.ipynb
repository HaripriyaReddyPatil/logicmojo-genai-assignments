## Problem 1: Names report


```python
names = ["Ada", "Alan", "Grace", "Kay", "Katherine", "Ed"]

long_names = 0

for i, name in enumerate(names, start=1):
    if len(name) > 4:
        print(f"{i}. {name} (long)")
        long_names += 1
    else:
        print(f"{i}. {name}")

print()
print(f"{long_names} of {len(names)} names are longer than four letters.")
```

## Problem 2: Reading signs

```python
readings = [12, -5, 0, 8, -3, 7]

positive_count = 0
total = 0

for value in readings:
    if value > 0:
        print(f"{value}: positive")
        positive_count += 1
    elif value < 0:
        print(f"{value}: negative")
    else:
        print(f"{value}: zero")

    total += value

print()
print(f"positives: {positive_count}")
print(f"total: {total}")
```

## Problem 3: Countdown list

```python
n = 5

countdown = []

while n > 0:
    countdown.append(n)
    n -= 1

print(countdown)
print(f"length: {len(countdown)}")
print(f"first: {countdown[0]}, last: {countdown[-1]}")
```

## Problem 4: Highest reading without max()

```python
scores = [12, 47, 47, 8, 5, 47, 19]

highest = scores[0]
first_index = 0

for i in range(len(scores)):
    if scores[i] > highest:
        highest = scores[i]
        first_index = i

count = 0

for score in scores:
    if score == highest:
        count += 1

print(f"highest: {highest}")
print(f"first seen at index: {first_index}")
print(f"appears {count} times")
```

## Problem 5: Is it sorted?

```python
values = [3, 8, 8, 12, 5, 20]

is_sorted = True
first_drop = -1

for i in range(len(values) - 1):
    if values[i] > values[i + 1]:
        is_sorted = False

        if first_drop == -1:
            first_drop = i

print(f"sorted: {is_sorted}")

if first_drop != -1:
    print(f"first drop at index {first_drop} ({values[first_drop]} then {values[first_drop + 1]})")
```
