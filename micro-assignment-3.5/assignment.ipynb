## Problem 1: Softmax with temperature

```python
import numpy as np

def softmax_T(logits, T):
    logits = np.array(logits, dtype=float)
    scaled = logits / T
    exp_values = np.exp(scaled)
    return exp_values / np.sum(exp_values)

probs = softmax_T([2, 1, 0], 1)

print("probs:", np.round(probs, 3))
```

## Problem 2: Lower temperature is sharper

```python
low_temp = softmax_T([2, 1, 0], 0.5)
high_temp = softmax_T([2, 1, 0], 2)

print(f"T=0.5 max: {np.max(low_temp):.2f}")
print(f"T=2   max: {np.max(high_temp):.2f}")
```

## Problem 3: Nucleus size for top-p

```python
probs = [0.5, 0.25, 0.15, 0.1]
p = 0.9

total = 0
count = 0

for prob in probs:
    total += prob
    count += 1

    if total >= p:
        break

print(f"nucleus size (p=0.9): {count}")
```

## Problem 4: Temperature 0 takes the top token

```python
logits = np.array([2., 1., 3.])

chosen_index = np.argmax(logits)

print("chosen index:", chosen_index)
```
