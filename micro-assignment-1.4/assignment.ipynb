```python
import numpy as np
import pandas as pd
```

## Problem 1: Array basics

```python
arr = np.array([2, 4, 6, 8, 10])

plus_ten = arr + 10
print("plus 10:", plus_ten)

first = np.array([1, 2, 3])
second = np.array([4, 5, 6])

product = first * second
print("product:", product)

print("mean:", arr.mean())
```

## Problem 2: Above the mean

```python
data = np.array([3, 8, 1, 10, 6, 4])

mean_value = data.mean()
mask = data > mean_value
above_mean = data[mask]

print(f"mean: {mean_value:.2f}")
print("above mean:", above_mean)
print("count:", len(above_mean))
```

## Problem 3: Vectorized equals loop

```python
qty = np.array([2, 1, 3])
price = np.array([10, 25, 5])

vectorized_total = np.sum(qty * price)

loop_total = 0

for i in range(len(qty)):
    loop_total += qty[i] * price[i]

print("vectorized:", vectorized_total)
print("loop:", loop_total)
print("match:", vectorized_total == loop_total)
```

## Problem 4: Normalize to 0..1

```python
scores = np.array([50, 20, 80, 40])

smallest = scores.min()
largest = scores.max()

normalized = (scores - smallest) / (largest - smallest)

print("normalized:", np.round(normalized, 2))
```

## Problem 5: A DataFrame

```python
df = pd.DataFrame({
    "name": ["Ada", "Alan", "Grace", "Kay"],
    "score": [91, 72, 85, 60],
    "team": ["A", "B", "A", "B"]
})

print(df)

print()
print("scores:", df["score"].tolist())

print()
high_scores = df[df["score"] >= 85]
print(high_scores)
```

## Problem 6: Derived column and group

```python
df = pd.DataFrame({
    "name": ["Ada", "Alan", "Grace", "Kay"],
    "score": [91, 72, 85, 60],
    "team": ["A", "B", "A", "B"]
})

df["passed"] = df["score"] >= 70

print(df)

print()

team_average = df.groupby("team")["score"].mean()
print(team_average)
```

## Problem 7: Clean a messy table

```python
raw = pd.DataFrame({
    "city": ["A", "B", "C", "D"],
    "sales": [100.0, np.nan, 250.0, 80.0],
    "visits": [10, 5, 25, 8]
})

clean = raw.dropna()
clean = clean[clean["sales"] > 90].copy()

clean["per_visit"] = clean["sales"] / clean["visits"]

print(clean)
```

## Problem 8: Top category

```python
sales = pd.DataFrame({
    "category": ["food", "toys", "food", "books", "toys"],
    "amount": [20, 15, 30, 10, 25]
})

totals = sales.groupby("category")["amount"].sum()

print(totals)

print()

top_category = totals.idxmax()
top_total = totals.max()

print(f"top category: {top_category} with {top_total}")
```
