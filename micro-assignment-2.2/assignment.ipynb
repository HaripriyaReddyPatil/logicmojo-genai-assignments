## Problem 1: Mean squared error by hand

```python
y_hat = [4.0, 1.0, 3.0, 6.0]
y = [3.5, 2.0, 3.0, 5.0]

squared_errors = []

for predicted, actual in zip(y_hat, y):
    error = predicted - actual
    squared_errors.append(error ** 2)

mse = sum(squared_errors) / len(squared_errors)

print(f"MSE: {mse:.2f}")
```

## Problem 2: One descent step

```python
theta = 0.8
grad = 2.0
eta = 0.1

theta_new = theta - eta * grad

print(f"theta_new: {theta_new:.1f}")
```

## Problem 3: A five-step descent

```python
x = 0.0
eta = 0.3

values = [x]

for step in range(5):
    grad = 2 * (x - 3)
    x = x - eta * grad
    values.append(round(x, 3))

print(f"x each step: {values}")
```

## Problem 4: Diagnose three learning rates

```python
learning_rates = [0.05, 0.4, 1.1]

labels = []

for eta in learning_rates:
    x = 1.0
    start_loss = x ** 2

    for step in range(15):
        grad = 2 * x
        x = x - eta * grad

    final_loss = x ** 2

    if final_loss > start_loss:
        labels.append("diverges")
    elif final_loss < 0.01:
        labels.append("converges")
    else:
        labels.append("crawls")

print(labels)
```

## Problem 6: Cross-entropy on a confident-wrong guess

```python
import math

p_right = 0.9
p_wrong = 0.1

loss_right = -math.log(p_right)
loss_wrong = -math.log(p_wrong)

print(f"cross-entropy at p=0.9: {loss_right:.2f}")
print(f"cross-entropy at p=0.1: {loss_wrong:.2f}")
print(
    "note: the confident-wrong guess costs far more, which is why classification uses cross-entropy"
)
```
