## Problem 1: A neuron forward pass

```python
import numpy as np

w = np.array([0.5, -0.2, 0.1])
x = np.array([2, 1, 4])
b = 0.3

z = np.dot(w, x) + b
relu = max(0, z)

print(f"z: {z:.1f}")
print(f"ReLU(z): {relu:.1f}")
```

## Problem 2: Apply ReLU, sigmoid, and tanh

```python
v = np.array([-2.0, 0.0, 3.0])

relu = np.maximum(0, v)
sigmoid = 1 / (1 + np.exp(-v))
tanh = np.tanh(v)

print("ReLU:", np.round(relu, 2))
print("sigmoid:", np.round(sigmoid, 2))
print("tanh:", np.round(tanh, 2))
```

## Problem 3: A layer of two neurons

```python
W = np.array([
    [0.5, -1.0],
    [1.0, 0.5]
])

b = np.array([0.0, 1.0])
x = np.array([2.0, 3.0])

z = W @ x + b
outputs = np.maximum(0, z)

print("outputs:", outputs)
```

## Problem 4: Two linear layers collapse to one

```python
W1 = np.array([
    [1.0, 2.0],
    [0.0, 1.0]
])

b1 = np.array([0.0, 0.0])

W2 = np.array([
    [2.0, -1.0]
])

b2 = np.array([0.0])

x = np.array([1.0, 1.0])

hidden = W1 @ x + b1
two_layers = W2 @ hidden + b2

combined_W = W2 @ W1
one_layer = combined_W @ x + b2

print("two layers:", two_layers)
print("one layer:", one_layer)
print("identical:", np.allclose(two_layers, one_layer))
```

## Problem 5: Softmax

```python
logits = np.array([1.0, 2.0, 3.0])

exp_values = np.exp(logits)
probs = exp_values / np.sum(exp_values)

print("probs:", np.round(probs, 3))
print("sum:", round(np.sum(probs), 1))
```

## Problem 6: Cross-entropy against a one-hot target

```python
y = np.array([0, 0, 1])
probs = np.array([0.1, 0.2, 0.7])

cross_entropy = -np.sum(y * np.log(probs))

p_true = probs[2]
same_value = -np.log(p_true)

print(f"cross-entropy: {cross_entropy:.2f}")
print(f"same as -log(p_true): {same_value:.2f}")
print(
    "note: cross-entropy grows toward infinity as the probability on the true class nears 0, so a confident wrong answer is punished hardest"
)
```
