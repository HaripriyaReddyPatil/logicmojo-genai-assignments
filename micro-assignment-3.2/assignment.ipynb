## Problem 1: A similarity score (dot product)

```python
import numpy as np

q = np.array([2., 0., 1.])
k = np.array([1., 3., 1.])

score = np.dot(q, k)

print("score:", score)
```

## Problem 2: Softmax over scores

```python
s = np.array([4., 1., 0.])

exp_scores = np.exp(s)
weights = exp_scores / np.sum(exp_scores)

print("weights:", np.round(weights, 3))
print("sum:", round(np.sum(weights), 1))
```

## Problem 3: Scale by sqrt(d_k)c

```python
scores = np.array([8., 4., 2.])
d_k = 4

scaled = scores / np.sqrt(d_k)

print("scaled:", scaled.tolist())
```

## Problem 4: The context vector (weighted blend)

```python
weights = np.array([0.9, 0.1])

values = np.array([
    [1., 0.],
    [0., 1.]
])

context = weights @ values

print("context:", context.tolist())
```

## Problem 5: Where does this token attend?

```python
row = np.array([0.1, 0.7, 0.2])

most_attended = np.argmax(row)

print("most-attended index:", most_attended)
```

## Problem 6: Why scale the scores?

```python
print(
    "dividing by sqrt(d_k) keeps dot products from growing large with vector length, "
    "so softmax stays smooth and training is stable"
)
```

## Problem 7: Causal mask (no peeking ahead)

```python
scores = np.array([2.0, 1.0, 3.0, 0.5])

masked_scores = scores.copy()

# Position 1 can only look at positions 0 and 1
masked_scores[2:] = -np.inf

exp_scores = np.exp(masked_scores)
weights = exp_scores / np.sum(exp_scores)

print("weights:", np.round(weights, 3))
```
