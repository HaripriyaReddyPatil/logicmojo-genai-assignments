## Problem 2: Three-way split and model selection

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

X, y = load_wine(return_X_y=True)

# First split: 80% train+validation, 20% test
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=0,
    stratify=y
)

# Second split: train and validation
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=0.25,
    random_state=0,
    stratify=y_train_val
)

# Scale using training data only
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

# Candidate 1
logreg = LogisticRegression(max_iter=5000)
logreg.fit(X_train_scaled, y_train)

# Candidate 2
tree = DecisionTreeClassifier(max_depth=2, random_state=0)
tree.fit(X_train_scaled, y_train)

logreg_val = logreg.score(X_val_scaled, y_val)
tree_val = tree.score(X_val_scaled, y_val)

print(
    f"sizes train/val/test: "
    f"{len(X_train)} {len(X_val)} {len(X_test)}"
)

print(f"validation logreg: {round(logreg_val, 2)}")
print(f"validation tree: {round(tree_val, 2)}")

if logreg_val > tree_val:
    chosen = "logreg"
    test_accuracy = logreg.score(X_test_scaled, y_test)
else:
    chosen = "tree"
    test_accuracy = tree.score(X_test_scaled, y_test)

print(f"chosen: {chosen}")
print(f"test accuracy: {round(test_accuracy, 2)}")
```

## Problem 6: Precision and recall on imbalanced labels

```python
labels = [0] * 18 + [1, 1]
preds = [0] * 20

correct = 0
true_positive = 0
false_positive = 0
false_negative = 0

for actual, predicted in zip(labels, preds):

    if actual == predicted:
        correct += 1

    if actual == 1 and predicted == 1:
        true_positive += 1

    elif actual == 0 and predicted == 1:
        false_positive += 1

    elif actual == 1 and predicted == 0:
        false_negative += 1


accuracy = correct / len(labels)

if true_positive + false_positive == 0:
    precision = 0.0
else:
    precision = true_positive / (true_positive + false_positive)

recall = true_positive / (true_positive + false_negative)

print(f"accuracy: {accuracy}")
print(f"precision: {precision}")
print(f"recall: {recall}")
print(
    "note: accuracy is high, but recall 0.0 means it caught none of the rare positives"
)
```
