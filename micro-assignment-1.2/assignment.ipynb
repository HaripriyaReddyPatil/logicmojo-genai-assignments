## Problem 1: Two small functions


```python
def rectangle_area(width: int, height: int) -> int:
    area = width * height
    return area

def greet(name: str, greeting: str = "Hello") -> str:
    message = f"{greeting}, {name}"
    return message

print(f"area of 3 x 4 = {rectangle_area(3, 4)}")
print(greet("Ada"))
print(greet("Alan", "Hi"))
```

## Problem 2: Clean a messy record

```python
raw = "  Ada Lovelace,  ADA@Math.ORG , London  "

parts = raw.split(",")

name = parts[0].strip()
email = parts[1].strip().lower()
city = parts[2].strip()

domain = email.split("@")[1]

name_parts = name.split()
initials = ""

for part in name_parts:
    initials += part[0].upper() + "."

print(f"name: {name}")
print(f"email: {email}")
print(f"city: {city}")
print(f"domain: {domain}")
print(f"initials: {initials}")
```

## Problem 3: Scores dictionary

```python
scores = {"Ada": 91, "Alan": 72, "Grace": 85}

total = 0

for name, score in scores.items():
    print(f"{name}: {score}")
    total += score

average = total / len(scores)

print(f"average: {average:.2f}")
```

## Problem 4: Keep the long words

```python
words = ["ai", "python", "ml", "agent", "rag"]

long_words = [word.upper() for word in words if len(word) > 2]

print(long_words)
print(f"{len(long_words)} words kept")
```

## Problem 5: Shared tags

```python
monday = ["python", "git", "venv", "ml"]
tuesday = ["git", "ml", "numpy", "pandas"]

monday_set = set(monday)
tuesday_set = set(tuesday)

shared = sorted(monday_set & tuesday_set)
all_tags = sorted(monday_set | tuesday_set)

print(f"in both: {shared}")
print(f"in either: {all_tags}")
print(f"{len(shared)} shared, {len(all_tags)} total")
```

## Problem 6: Longest name, without max()

```python
names = ["Ada", "Alan", "Grace", "Katherine", "Ed"]

lengths = {name: len(name) for name in names}

print(lengths)

longest = names[0]

for name in names:
    if len(name) > len(longest):
        longest = name

print(f"longest: {longest} ({len(longest)} letters)")
```
